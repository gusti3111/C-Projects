Faktor dari bilangan X


```python
x = int(input("Masukkan nilai x: "))
fact = 0
notFact =0
for f in range(1,20):
  if x % f == 0:
     print(str(f) + " merupakan faktor dari " + str(x))
     fact +=1
  else:
    print(str(f) + " bukan merupakan faktor dari bilangan " + str(x))
    notFact +=1

print(str(fact) + " is factor of " + str(x))
print(str(notFact)+ " is not factor of "+ str(x))

```

email validation function


```python
from email_validator import validate_email, EmailNotValidError

def check(email):
    try:
      # validate and get info
        v = validate_email(email)
        # replace with normalized form
        email = v["email"]
        print("True")
    except EmailNotValidError as e:
        # email is not valid, exception message is human-readable
        print(str(e))

check("tommy.whiteranger@gmail.com")

check("ankitrai326.com")
```
